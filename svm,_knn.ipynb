```python
class SVM():
  def __init__(self, lr, no_iters, lmb):
    self.lr = lr
    self.no_iters = no_iters
    self.lmb = lmb

  def fit(self, X, y):
    self.m, self.n = X.shape
    self.w = np.zeroes(self.n)
    self.b = 0
    self.X = X
    self.y = y
    for i in range(self.no_iters):
      self.update()

  def update(self):
    label = np.where(self.y <=0, -1, 1)
    for idx, xi in enumerate(self.X):
      condition = label[idx]*(np.dot(xi, self.w) - self.b) >= 1
      if condition:
        dw = 2 * self.lmb * self.w
        db = 0
      else:
        dw = 2 * self.lmb * self.w - np.dot(label[idx], xi)
        db = label[idx]
      self.w -= self.lr * dw
      self.b -= self.lr * db

  def predict(self, X):
    output = np.dot(self.X, self.w) - self.b
    pred = np.sign(output)
    return np.where(pred <=-1 , 0, 1)
```

```python
SVM = SVM(0.01, 1000, 0.1)
```

```python
class KNN():
  def __init__(self, dist):
    self.distance_metric = dist
  def distance_metric(self, training, testing):
    d = 0
    if self.distance_metric == 'Euclidean':
      for i in range(len(training)-1):
        d += (training[i] - testing[i])**2
      return d
    elif self.distance_metric == 'Manhattan':
      for i in range(len(training)-1):
        d += abs(training[i] - testing[i])
      return d
  def nearest(self, X_train, testing, k):
    dlist = []
    for i in X_train:
      d = self.distance_metric(i, testing)
      dlist.append(i, d)
    dlist.sort(key = lambda x: x[1])
    nlist = []
    for j in range(k):
      nlist.append(dlist[j][0])
    return nlist
```
