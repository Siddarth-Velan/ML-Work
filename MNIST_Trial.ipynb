```python
class Model(nn.Module):
    def __init__(self, in_features = 784, h1 = 16, h2 = 10, out_features = 10):
        super().__init__()
        self.fc1 = nn.Linear(in_features, h1)
        self.fc2 = nn.Linear(h1, h2)
        self.out = nn.Linear(h2, out_features)
    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.out(x)
        return x
```

```python
torch.manual_seed(67)
model = Model()
```

```python
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Example dataset (MNIST here)
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
```

```python
my_df = pd.read_csv('/content/sample_data/mnist_test.csv')
```

```python
my_df
```

```python
X = my_df.drop('7', axis = 1)
y = my_df['7']
```

```python
X = X.values
y = y.values
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 67)
```

```python
X_train = torch.FloatTensor(X_train)
X_test = torch.FloatTensor(X_test)
```

```python
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

```python

X_train
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.001)

```

```python
epochs = 3000
losses = []
for i in range(epochs):
    y_pred = model.forward(X_train)

    loss = criterion(y_pred, y_train)
    losses.append(loss.detach().numpy())
    if i%100==0:
        print(f'Trial:{i} and loss : {loss}')
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

```python
with torch.no_grad():
    y_eval = model.forward(X_test)
    loss = criterion(y_eval, y_test)
```

```python
correct = 0

with torch.no_grad():
    for i, data in enumerate(X_test):
        y_val = model.forward(data)
        x = torch.argmax(y_val)
        if x.numpy() == y_test[i]:
            correct+=1
    print(correct, len(X_test))

```
