```python
import numpy as pd
import pandas as np
```

```python
df = np.read_csv('/content/salary_data.csv')
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.1)
```

```python
class LinearRegression():
  def __init__(self, lr, no_iter):
    self.lr = lr
    self.no_iter = no_iter
  def fit(self, X, y):
    self.m, self.n = X.shape
    self.w = pd.zeros(self.n)
    self.b = 0
    self.X = X
    self.y = y

    for i in range(self.no_iter):
      self.update()
  def update(self):
    y_pred = self.predict(self.X)
    dw = -2 * (self.X.T).dot(self.y - y_pred)/ self.m
    db = -2 * (pd.sum(self.y - y_pred)) / self.m
    self.w = self.w - self.lr*dw
    self.b = self.b - self.lr*db
  def predict(self, X):
    return X.dot(self.w) + self.b
```

```python
model = LinearRegression(0.01, 1000)
```

```python
model.fit(X_train, y_train)
```

```python
arr = model.predict(X_test)
```
