```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
from collections import Counter
```

```python
def PCA(X, k):
  X_c = X - X.mean(axis = 0)
  cov = np.cov(X_c, rowvar = False)
  eigvals, eigvecs = np.linalg.eigh(cov)
  idx = np.argsort(eigvals)[::-1]
  eigvals = eigvals[idx]
  eigvecs = eigvecs[:, idx]
  V_k = eigvecs[:, :k]
  Z = X_c @ V_k
  return Z
```

```python
class KNN():
    def __init__(self, dist="Euclidean"):
        self.dist_name = dist

    def distance(self, a, b):
        if self.dist_name == "Euclidean":
            return sum((a[i] - b[i])**2 for i in range(len(a)))
        elif self.dist_name == "Manhattan":
            return sum(abs(a[i] - b[i]) for i in range(len(a)))
        else:
            raise ValueError("Unknown distance metric")

    def nearest(self, X_train, y_train, x_test, k):
        dlist = []
        for i in range(len(X_train)):
            d = self.distance(X_train[i], x_test)
            dlist.append((d, y_train[i]))

        dlist.sort(key=lambda x: x[0])

        neighbors = [label for _, label in dlist[:k]]
        return neighbors

    def predict_one(self, X_train, y_train, x_test, k):
        neighbors = self.nearest(X_train, y_train, x_test, k)
        return max(set(neighbors), key=neighbors.count)

    def predict(self, X_train, y_train, X_test, k):
        return [self.predict_one(X_train, y_train, x_test, k) for x_test in X_test]
```

```python
def accuracy(y_test, y_pred):
  return np.sum(y_test == y_pred) / len(y_test)
```

```python
data = datasets.load_iris()
X, y = data.data, data.target
X_r = PCA(X, 2)
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(X_r, y)
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
knn = KNN('Euclidean')
pred1 = knn.predict(X_train, y_train, X_test, k=5)
pred2 = knn.predict(X_train_2, y_train_2, X_test_2, k=5)
```

```python
accuracy(y_test, pred1) , accuracy(y_test_2, pred2)
```
